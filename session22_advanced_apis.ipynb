```python
# Task 1: Demonstrate cookie persistence across requests using a Session object
import requests

# Creating a stateful session instance
session = requests.Session()

# 1st Request: Tell the server to set a target session cookie
print("--- Sending Request 1 (Setting Session Cookie) ---")
set_cookie_url = "https://httpbin.org/cookies/set/session_user/rupanshu29"
response1 = session.get(set_cookie_url)
print(f"Request 1 Status: {response1.status_code}")

# 2nd Request: Hit the cookie index to check if the session object kept it alive
print("\n--- Sending Request 2 (Reading Sent Cookies) ---")
check_cookies_url = "https://httpbin.org/cookies"
response2 = session.get(check_cookies_url)
print(f"Request 2 Status: {response2.status_code}")
print(f"Session Cookies Remembered: {response2.json()}")
```

```python
# Task 2: Fetch current temperature for Ahmedabad using OpenWeatherMap API
import requests

url = "https://api.openweathermap.org/data/2.5/weather"

# Replace this string with your actual OpenWeatherMap API key
API_KEY = "YOUR_API_KEY_HERE"

query_params = {
    "q": "Ahmedabad",
    "appid": API_KEY,
    "units": "metric" # Returns temperature value readings directly in Celsius
}

response = requests.get(url, params=query_params)

if response.status_code == 200:
    weather_data = response.json()
    temp = weather_data["main"]["temp"]
    condition = weather_data["weather"][0]["description"]
    print(f"--- Current Weather in Ahmedabad ---")
    print(f"Temperature: {temp}Â°C")
    print(f"Condition  : {condition.title()}")
elif response.status_code == 401:
    print("API Error: Unauthorized. Please check if your 'API_KEY' is valid and active.")
else:
    print(f"Failed to fetch data. Server Status Code: {response.status_code}")
```

```python
# Task 3: Simulate fetching multiple API data pipelines concurrently
# Install async-compatible HTTP library if missing from local kernel
!pip install httpx

import asyncio
import httpx

async def fetch_mock_data(client, source_name, target_url):
    print(f"[Start] Fetching from {source_name}...")
    response = await client.get(target_url)
    print(f"[Done] Received response from {source_name}.")
    return response.json()

async def main():
    # Public non-blocking endpoints used for async simulation modeling
    spotify_mock_url = "https://jsonplaceholder.typicode.com/posts/1"
    bookmyshow_mock_url = "https://jsonplaceholder.typicode.com/posts/2"
    
    async with httpx.AsyncClient() as client:
        # Schedule both network operations to run concurrently using gather()
        tasks = [
            fetch_mock_data(client, "Spotify Trends", spotify_mock_url),
            fetch_mock_data(client, "BookMyShow Movies", bookmyshow_mock_url)
        ]
        
        print("--- Starting Concurrent API Requests ---")
        results = await asyncio.gather(*tasks)
        print("--- All Data Fetched Successfully ---\n")
        
        print(f"Result 1 (Spotify Mock Title): {results[0].get('title')}")
        print(f"Result 2 (BookMyShow Mock Body): {results[1].get('body')[:50]}...")

# Entry point handling execution hooks inside active notebook event loops
await main()
```

```python
# Task 4: Call a protected API endpoint passing an Authorization Bearer token
import requests

url = "https://jsonplaceholder.typicode.com/users/1"

# Simulating an access token generated from a login system
mock_bearer_token = "YOUR_TOKEN_HERE"

# Enforcing constraint format requirements: 'Bearer <token>'
request_headers = {
    "Authorization": f"Bearer {mock_bearer_token}",
}

response = requests.get(url, headers=request_headers)

if response.status_code == 200:
    profile_data = response.json()
    print("--- User Profile Retrieved Successfully ---")
    print(f"User Name: {profile_data.get('name')}")
    print(f"Company  : {profile_data.get('company', {}).get('name')}")
else:
    print(f"Authorization Rejected. Status Code: {response.status_code}")
```

```python
# Task 5: Generate the starting URL redirect sequence for an OAuth 2.0 Auth Flow
import urllib.parse

def generate_spotify_oauth_url():
    base_auth_url = "https://accounts.spotify.com/authorize"
    
    # Query string parameters required to verify client app initialization steps
    oauth_params = {
        "client_id": "mock_client_id_987654321",
        "response_type": "code",
        "redirect_uri": "https://localhost:8080/callback",
        "scope": "user-read-private user-read-email playlist-modify-public"
    }
    
    # Safe encoding helper to convert map parameters into valid URL entities
    encoded_query_string = urllib.parse.urlencode(oauth_params)
    complete_redirect_url = f"{base_auth_url}?{encoded_query_string}"
    
    return complete_redirect_url

# Generate and view url sequence strings
redirect_target = generate_spotify_oauth_url()

print("--- Generated OAuth 2.0 Authorization Landing URL ---")
print(redirect_target)

print("\n--- Summary of what this URL does ---")
print("1. The app redirects the user to this generated Spotify login interface address.")
print("2. The user reviews the requested application permissions ('scopes') and logs in securely.")
print("3. Upon authorization, Spotify sends a unique access grant 'code' back to the 'redirect_uri'.")
```
