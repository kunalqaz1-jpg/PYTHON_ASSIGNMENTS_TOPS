```python
# Task 1: Import math module and calculate the square root of 225
import math

number = 225
square_root = math.sqrt(number)

print(f"The square root of {number} is: {square_root}")
```

```python
# Task 2: Create a folder named 'MyDownloads' and get its absolute path
import os

folder_name = 'MyDownloads'

# exist_ok=True prevents errors if you run this cell multiple times
os.makedirs(folder_name, exist_ok=True)

# Fetching the absolute path of the newly created folder
absolute_path = os.path.abspath(folder_name)

print(f"Folder '{folder_name}' created successfully!")
print(f"Absolute Path: {absolute_path}")
```
