```python
# Task 1: Function calculating average duration per song with explicit ZeroDivisionError tracking

def get_song_duration_per_minute(total_duration, total_songs):
    try:
        result = total_duration / total_songs
        print(f"Average duration per song: {result:.2f} minutes")
    except ZeroDivisionError:
        print("Error: Total songs cannot be zero!")
    finally:
        print("Calculation complete!\n")

# --- Demonstration ---
get_song_duration_per_minute(60, 5)
get_song_duration_per_minute(60, 0)
```

```python
# Task 2: Evaluate price per item with multiple exception guard conditions

try:
    total = float(input("Enter total cart amount: "))
    count = int(input("Enter number of items: "))
    
    price_per_item = total / count
    print(f"Price per item: ₹{price_per_item:.2f}")

except ZeroDivisionError:
    print("Error: Item count cannot be zero!")
except ValueError:
    print("Error: Invalid numeric input entered!")
```

```python
# Task 3: Defining and raising a custom exception class 'NoOffersApplied'

class NoOffersApplied(Exception):
    """Custom exception raised when an order has zero active coupon reductions."""
    pass

try:
    spend = float(input("Enter total spend: "))
    offers = int(input("Enter number of offers: "))
    
    if offers == 0:
        raise NoOffersApplied("No offers were applied!")
        
    print(f"Average cashback per offer: ₹{spend / offers:.2f}")

except NoOffersApplied as e:
    print(f"Custom Error: {e}")
except ZeroDivisionError:
    print("Error: Offers cannot be zero if evaluating cashback partitions!")
```

```python
# Task 4: Movie rating calculator utilizing error-safe tracking architectures

def calculate_average_rating(total_rating, num_reviews):
    try:
        result = total_rating / num_reviews
        print(f"Average Rating: {result:.1f}/5")
    except ZeroDivisionError:
        print("Error: No reviews found!")
    finally:
        print("Thank you for using the calculator")

# --- Demonstration ---
calculate_average_rating(500, 0)
```

```python
# Task 5: Bill splitting mechanism utilizing an else conditional clause block

def safe_divide_for_zomato(bill_amount, num_people):
    try:
        result = bill_amount / num_people
    except ZeroDivisionError:
        print("Error: Number of people cannot be zero!")
    else:
        # Executes ONLY if no exceptions were raised in the try block
        print(f"Each person pays: ₹{result:.2f}")
    finally:
        print("Split calculation done\n")

# --- Demonstration ---
print("--- Split 1 ---")
safe_divide_for_zomato(800, 4)

print("--- Split 2 ---")
safe_divide_for_zomato(800, 0)
```
