```python
# Tasks 1, 2 & 5: Creating an evolved Song class with optional duration and preview method

class Song:
    # Task 5: Use a default argument for duration (default to 0 if not provided)
    def __init__(self, title, artist, duration=0):
        self.title = title
        self.artist = artist
        self.duration = duration  # Duration stored in seconds

    # Task 2: Method to play a 30-second preview of the song
    def play_preview(self):
        print(f"Playing 30-second preview of '{self.title}' by {self.artist}")

# --- Demonstration ---

# Task 1 & 5: Instantiating an object for a favorite song (with duration provided)
favorite_song = Song(title="Starboy", artist="The Weeknd", duration=230)

print("--- Song Details ---")
print(f"Title: {favorite_song.title}")
print(f"Artist: {favorite_song.artist}")
print(f"Duration: {favorite_song.duration} seconds\n")

# Task 2: Call the preview method
print("--- Triggering Preview ---")
favorite_song.play_preview()

# Verification of Task 5 (Optional parameter behavior when omitted):
print("\n--- Testing Optional Parameter ---")
hidden_track = Song(title="Mystery Track", artist="Unknown Artist")
print(f"Track: {hidden_track.title}, Default Duration: {hidden_track.duration} seconds")
```

```python
# Tasks 3 & 4: FoodOrder class managing attributes and real-time total price updates

class FoodOrder:
    # Task 3: Constructor initializing restaurant details, item list, and total bill
    def __init__(self, restaurant_name, items=None, total_price=0.0):
        self.restaurant_name = restaurant_name
        # If no initial list is passed, initialize an empty list safely
        self.items = items if items is not None else []
        self.total_price = float(total_price)

    # Task 4: Method to add a single item and update the running total price
    def add_item(self, item_name, item_price):
        self.items.append(item_name)
        self.total_price += float(item_price)
        print(f"Added: {item_name} (+₹{item_price})")

# --- Demonstration ---

# Task 3: Simulating a recent initial order state
my_order = FoodOrder(
    restaurant_name="KFC", 
    items=["Zinger Burger", "Fries"], 
    total_price=350.0
)

print("--- Initial Zomato/Swiggy Order ---")
print(f"Restaurant: {my_order.restaurant_name}")
print(f"Items Ordered: {my_order.items}")
print(f"Current Total Bill: ₹{my_order.total_price}\n")

# Task 4: Simulating extending the order dynamically by adding two extra items
print("--- Modifying Order ---")
my_order.add_item(item_name="Chiken Popcorn", item_price=120.0)
my_order.add_item(item_name="Pepsi Can", item_price=60.0)

print("\n--- Final Updated Invoice ---")
print(f"Restaurant: {my_order.restaurant_name}")
print(f"Final Items List: {my_order.items}")
print(f"Final Updated Total Bill: ₹{my_order.total_price}")
```
