```python
# Task 1: Fetch posts, print response status code and the first post's title
import requests

url = "https://jsonplaceholder.typicode.com/posts"
response = requests.get(url)

print(f"Status Code: {response.status_code}")

if response.status_code == 200:
    posts = response.json()
    # Get the title field from the dictionary at index 0
    first_post_title = posts[0]['title']
    print(f"First Post Title: {first_post_title}")
else:
    print("Failed to retrieve posts.")
```

```python
# Task 2: Send a POST request with specific payload details
import requests

url = "https://jsonplaceholder.typicode.com/posts"

post_data = {
    "title": "My First Post",
    "body": "Hello from Python!",
    "userId": 101
}

# Sending payload as a JSON object
response = requests.post(url, json=post_data)

print(f"Status Code: {response.status_code}")
print("--- Returned JSON Response ---")
print(response.json())
```

```python
# Task 3: Fetch all users and filter usernames whose email ends with '.org'
import requests

url = "https://jsonplaceholder.typicode.com/users"
response = requests.get(url)

if response.status_code == 200:
    users = response.json()
    
    print("--- Usernames with .org Emails ---")
    for user in users:
        email = user.get('email', '')
        # Filter strings dynamically regardless of casing issues
        if email.lower().endswith('.org'):
            print(f"User: {user['username']} | Email: {email}")
else:
    print(f"Failed to fetch users. Status Code: {response.status_code}")
```

```python
# Task 4: Search OMDB API using query parameters and print total results count
import requests

url = "http://www.omdbapi.com/"

# Configuring required parameters for search operation
query_params = {
    "apikey": "demo",
    "s": "Avengers"
}

response = requests.get(url, params=query_params)

if response.status_code == 200:
    movie_data = response.json()
    
    # Check if the API returned a successful data response flag
    if movie_data.get("Response") == "True":
        total_results = movie_data.get("totalResults")
        print(f"Search Query: 'Avengers'")
        print(f"Total Results Found: {total_results}")
    else:
        # Handle cases where the 'demo' API key might be expired/blocked
        print(f"OMDB API Error Message: {movie_data.get('Error', 'Unknown Error')}")
else:
    print(f"HTTP Request Connection Error. Status Code: {response.status_code}")
```
