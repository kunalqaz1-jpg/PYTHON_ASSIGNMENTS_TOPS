```python
#task3
import csv

# First creating the CSV file
with open('ipl_matches.csv', 'w', newline='') as f:
    writer = csv.writer(f)
    writer.writerow(['match_id', 'team1', 'team2', 'winner'])
    writer.writerow([1, 'CSK', 'MI', 'CSK'])
    writer.writerow([2, 'RCB', 'KKR', 'KKR'])
    writer.writerow([3, 'DC', 'SRH', 'SRH'])

# Now reading and printing winners
with open('ipl_matches.csv', 'r') as f:
    reader = csv.DictReader(f)
    for row in reader:
        print("Winner:", row['winner'])
```

```python
#task4
import json

movies = [
    {"title": "Pathaan", "year": 2023, "rating": 8.5},
    {"title": "RRR", "year": 2022, "rating": 9.0},
    {"title": "KGF2", "year": 2022, "rating": 8.8}
]

with open('movies.json', 'w') as f:
    json.dump(movies, f)

with open('movies.json', 'r') as f:
    data = json.load(f)
    for movie in data:
        print(movie['title'], "-", movie['rating'])
```
