```python
#task2
steps = [7000, 8500, 11000, 9500, 10500, 6000, 12000]

i = 0
while i < len(steps):
    if steps[i] > 10000:
        print("First day crossing 10,000 steps: Day", i+1, "with", steps[i], "steps")
        break
    i += 1
    
```

```python
#task3
def print_long_teams(teams):
    for team in teams:
        if len(team) <= 6:
            continue
        print(team)

ipl_teams = ["CSK", "Mumbai Indians", "RCB", "KKR", "Rajasthan Royals"]
print_long_teams(ipl_teams)
```

```python
#task4
durations = [210, 195, 240, 180, 220]

for i, duration in enumerate(durations, 1):
    print(f"Song {i}: {duration} seconds")
```

```python
#task5
prices = [500, 0, 800, 300, 0, 600, 900]

total = 0
for price in prices:
    if price == 0:
        continue
    total += price
    if total > 2000:
        break

print("Final Total: ₹", total)
```
