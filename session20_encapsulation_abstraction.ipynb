```python
# Task 1: Playlist class with a protected/private list attribute

class Playlist:
    def __init__(self):
        self._songs = []  # Private/Protected attribute by convention

    def add_song(self, song):
        """Public method to add a song to the private list."""
        self._songs.append(song)

    def get_playlist_tracks(self):
        """Helper method to safely return the list of tracks."""
        return self._songs

# --- Demonstration ---
my_playlist = Playlist()

# Adding 3 songs
my_playlist.add_song("Espresso")
my_playlist.add_song("Cruel Summer")
my_playlist.add_song("Birds of a Feather")

print("--- Current Playlist Tracks ---")
print(my_playlist.get_playlist_tracks())
```

```python
# Task 2: Flipkart-style Product class with explicit getter and setter methods

class Product:
    def __init__(self, name, price):
        self.name = name
        self._price = price  # Encapsulated attribute

    # Getter method
    def get_price(self):
        return self._price

    # Setter method
    def set_price(self, new_price):
        if new_price >= 0:
            self._price = new_price
        else:
            print("Error: Price cannot be negative!")

# --- Demonstration ---
product = Product("Wireless Headphones", 2499.00)

print("--- Initial Product Details ---")
print(f"Product: {product.name}")
print(f"Price: ₹{product.get_price()}")

# Updating the price using the setter
print("\nUpdating price due to sale...")
product.set_price(1999.00)
print(f"New Price: ₹{product.get_price()}")
```

```python
# Task 3: Movie class validating movie ratings through a setter method

class Movie:
    def __init__(self, title, rating):
        self.title = title
        self._rating = 0.0
        self.set_rating(rating) # Pass through the setter to validate initial setup

    # Getter method
    def get_rating(self):
        return self._rating

    # Setter method with a 0 to 10 validation constraint
    def set_rating(self, new_rating):
        if 0.0 <= new_rating <= 10.0:
            self._rating = float(new_rating)
        else:
            print(f"Error: Invalid rating ({new_rating}). Rating must be between 0 and 10.")

# --- Demonstration ---
movie = Movie("Inception", 8.8)

print("--- Movie Info ---")
print(f"Movie: {movie.title}")
print(f"Rating: {movie.get_rating()}/10")

# Testing invalid setter input
print("\n--- Attempting Invalid Update ---")
movie.set_rating(12.5)  # Out of bounds

# Verifying rating remained unchanged
print(f"Current safe rating: {movie.get_rating()}/10")
```

```python
# Task 4: Abstract class PaymentMethod with Paytm and PhonePe concrete subclasses
from abc import ABC, abstractmethod

# Defining the Abstract Base Class (ABC)
class PaymentMethod(ABC):
    
    @abstractmethod
    def pay(self, amount):
        """Abstract method that must be overridden by all subclasses."""
        pass

# Paytm implementation subclass
class Paytm(PaymentMethod):
    def pay(self, amount):
        print(f"Processing secure payment of ₹{amount} via Paytm Gateway.")

# PhonePe implementation subclass
class PhonePe(PaymentMethod):
    def pay(self, amount):
        print(f"Initiating fast payment of ₹{amount} using PhonePe UPI.")

# --- Demonstration ---
print("--- Simulating Checkout Steps ---")

paytm_wallet = Paytm()
phonepe_upi = PhonePe()

# Executing uniform interface actions across varying logic blocks
paytm_wallet.pay(450.00)
phonepe_upi.pay(1250.50)
```
