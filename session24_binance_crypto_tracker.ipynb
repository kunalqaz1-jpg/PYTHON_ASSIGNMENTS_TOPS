```python
# Tasks 1 to 4: Fetch Top 10 Cryptocurrencies, display 24h metrics, handle connection errors, and save to CSV
import requests
import csv

# Target CoinGecko markets API endpoint provided in instructions
url = "https://api.coingecko.com/api/v3/coins/markets?vs_currency=usd&order=market_cap_desc&per_page=10&page=1"

try:
    print("Connecting to CoinGecko Data Engine...")
    response = requests.get(url, timeout=10)
    
    # Task 4: Explicitly check HTTP response status codes before attempting to parse data
    if response.status_code == 200:
        crypto_list = response.json()
        
        # Open local CSV file pointer for writing data matrix records
        # newline="" prevents blank row spacing insertions inside Windows files
        with open("crypto_prices.csv", mode="w", encoding="utf-8", newline="") as csv_file:
            writer = csv.writer(csv_file)
            
            # Writing the column headers matching Task 3 requirement
            writer.writerow(["Name", "Current Price (USD)", "24h Change %", "24h High", "24h Low"])
            
            print("\n=====================================================================================")
            print(f"{'CRYPTO NAME':<18} | {'PRICE (USD)':<12} | {'24H CHANGE %':<12} | {'24H HIGH':<10} | {'24H LOW':<10}")
            print("=====================================================================================")
            
            # Tasks 1 & 2: Loop through top 10 coins and pull individual market fields
            for coin in crypto_list:
                name = coin.get("name")
                price = coin.get("current_price")
                change_24h = coin.get("price_change_percentage_24h")
                high_24h = coin.get("high_24h")
                low_24h = coin.get("low_24h")
                
                # Format variables cleanly for terminal display output
                change_str = f"{change_24h:+.2f}%" if change_24h is not None else "N/A"
                
                print(f"{name:<18} | ${price:<11,} | {change_str:<12} | ${high_24h:<9,} | ${low_24h:<9,}")
                
                # Task 3: Write extracted currency row values directly out to csv rows
                writer.writerow([name, price, change_24h, high_24h, low_24h])
                
        print("=====================================================================================")
        print(" Success! Live tracker stats written safely to 'crypto_prices.csv'.")
        
    elif response.status_code == 429:
        print("\n[API Rate Limit Exception]: You've made too many requests in a short period. Please wait a minute before running this cell again.")
    else:
        print(f"\n[Server Error]: API responded with status code {response.status_code}. Unable to read market parameters.")

# Task 4: Intercept unexpected lower-level networking dropouts or connection timeout faults
except requests.exceptions.RequestException as error_context:
    print(f"\n[Network Connection Exception]: Failed to communicate with CoinGecko.")
    print(f"Diagnostic Context: {error_context}")
```
