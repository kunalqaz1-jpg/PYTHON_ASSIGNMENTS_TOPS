```python
#task5
def get_unique_artists(spotify_playlist1, spotify_playlist2):
    return spotify_playlist1 | spotify_playlist2

playlist1 = {"Arijit Singh", "Shreya Ghoshal", "KK"}
playlist2 = {"Arijit Singh", "Neha Kakkar", "Atif Aslam"}

print(get_unique_artists(playlist1, playlist2))
```
