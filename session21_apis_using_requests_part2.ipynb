```python
# Task 1: Send a basic POST request with a title, body, and userId
import requests

url = "https://jsonplaceholder.typicode.com/posts"

payload = {
    "title": "Learning API Basics",
    "body": "Sending structured POST request bodies with Python Requests is simple.",
    "userId": 55
}

# Sending using the json= parameter ensures appropriate content headers are auto-attached
response = requests.post(url, json=payload)

print(f"Response Status Code: {response.status_code}")
print("\n--- Returned JSON Data From Server ---")
print(response.json())
```

```python
# Task 2: Prompt user for details and submit to a mock API endpoint
import requests

url = "https://jsonplaceholder.typicode.com/posts"

print("--- Create a New Spotify Playlist ---")
playlist_name = input("Enter playlist name: ")
playlist_desc = input("Enter playlist description: ")

# Building dynamic object payload structure
user_payload = {
    "title": playlist_name,
    "body": playlist_desc,
    "userId": 99
}

response = requests.post(url, json=user_payload)

if response.status_code == 201:
    returned_data = response.json()
    # Mock servers return a auto-incrementing ID to show registration success
    print(f"\n Success! Playlist created with ID: {returned_data.get('id')}")
else:
    print(f"Failed to create playlist. Status: {response.status_code}")
```

```python
# Task 3: POST user details to ReqRes and extract user ID and creation timestamp
import requests

url = "https://reqres.in/api/users"

user_profile = {
    "name": "Rupanshu",
    "job": "Python Developer"
}

response = requests.post(url, json=user_profile)

if response.status_code == 201:
    data = response.json()
    
    # Parsing specific return properties out of the response body dictionary
    user_id = data.get("id")
    created_at = data.get("createdAt")
    
    print("--- User Registration Details ---")
    print(f"Successfully Created User ID: {user_id}")
    print(f"Server Creation Timestamp  : {created_at}")
else:
    print(f"Request failed. Status Code: {response.status_code}")
```

```python
# Task 4: Fetch posts, extract the latest 5 elements, and export to posts.csv
import requests
import csv

url = "https://jsonplaceholder.typicode.com/posts"
response = requests.get(url)

if response.status_code == 200:
    all_posts = response.json()
    
    # Slicing the last 5 indices to get the latest posts uploaded
    latest_five_posts = all_posts[-5:]
    
    # Writing structured content rows into flat CSV schemas
    with open("posts.csv", mode="w", encoding="utf-8", newline="") as csv_file:
        writer = csv.writer(csv_file)
        
        # Writing the header row
        writer.writerow(["Title", "User ID"])
        
        # Iterating through arrays and logging specific map parameters
        for post in latest_five_posts:
            writer.writerow([post["title"], post["userId"]])
            
    print(" Successfully saved the latest 5 posts to 'posts.csv'!")
else:
    print(f"Unable to read external API server. Status: {response.status_code}")
```

```python
# Task 5: Save the same latest 5 posts dataset into a JSON file format
import requests
import json

url = "https://jsonplaceholder.typicode.com/posts"
response = requests.get(url)

if response.status_code == 200:
    all_posts = response.json()
    latest_five_posts = all_posts[-5:]
    
    # Open local file buffer to stream nested object trees directly
    with open("posts.json", mode="w", encoding="utf-8") as json_file:
        # indent=4 creates beautiful structured layout design rules
        json.dump(latest_five_posts, json_file, indent=4)
        
    print(" Successfully saved the latest 5 posts to 'posts.json'!")
else:
    print(f"Unable to read external API server. Status: {response.status_code}")
```
