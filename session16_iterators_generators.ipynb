```python
# Task 1: Manually loop through food delivery apps using iter() and next()

apps = ["Zomato", "Swiggy", "Domino's", "Blinkit", "Zepto"]

# Convert the list into an iterator
apps_iterator = iter(apps)

print("--- Food Delivery Apps ---")
# Using a loop to manually call next() until StopIteration is raised
while True:
    try:
        app = next(apps_iterator)
        print(app)
    except StopIteration:
        # Gracefully exit the loop when there are no items left
        break
```

```python
import random

# Task 2: Generator function simulating a Spotify playlist shuffle
def playlist_generator(songs_list):
    # Create a copy and shuffle it to simulate a realistic shuffle
    shuffled_playlist = songs_list.copy()
    random.shuffle(shuffled_playlist)
    
    for song in shuffled_playlist:
        yield song

# Demo usage:
my_songs = ["Song A", "Song B", "Song C", "Song D", "Song E"]
shuffle_player = playlist_generator(my_songs)

print("--- Shuffled Playlist ---")
for track in shuffle_player:
    print(f"Playing: {track}")
```

```python
# Task 3: Use enumerate() to display items and their item numbers

cart = ['Pizza', 'Burger', 'Fries', 'Coke']

print("--- Your Flipkart Cart ---")
# start=1 makes it display user-friendly numbering instead of 0-indexing
for index, item in enumerate(cart, start=1):
    print(f"Item {index}: {item}")
```

```python
# Task 4: Pair cricket teams with their IPL points using zip()

teams = ["Mumbai Indians", "Chennai Super Kings", "Royal Challengers Bengaluru", "Kolkata Knight Riders"]
points = [18, 16, 14, 20]

print("--- IPL Standings ---")
for team, point in zip(teams, points):
    print(f"Team: {team}, Points: {point}")
```

```python
# Task 5: Unique Order ID generator starting from 1001

def order_id_generator():
    current_id = 1001
    while True:
        yield current_id
        current_id += 1

# Demo usage:
id_gen = order_id_generator()

print("--- Generating Order IDs ---")
print(f"New Order Placed! ID: {next(id_gen)}")
print(f"New Order Placed! ID: {next(id_gen)}")
print(f"New Order Placed! ID: {next(id_gen)}")
```
