```python
#task1
def get_song_duration_per_minute(total_duration, num_songs):
    try:
        result = total_duration / num_songs
    except ZeroDivisionError:
        print("Error: No songs in playlist!")
    else:
        print("Average duration per song:", result, "minutes")
    finally:
        print("Calculation complete!")

get_song_duration_per_minute(60, 5)
get_song_duration_per_minute(60, 0)
```

```python
#task2
try:
    total = float(input("Enter total cart amount: "))
    count = int(input("Enter number of items: "))
    price_per_item = total / count
    print("Price per item: ₹", price_per_item)
except ZeroDivisionError:
    print("Error: Item count cannot be zero!")
```

```python
#task3
class NoOffersApplied(Exception):
    pass

try:
    spend = float(input("Enter total spend: "))
    offers = int(input("Enter number of offers: "))
    if offers == 0:
        raise NoOffersApplied("No offers were applied!")
    print("Average cashback per offer:", spend / offers)
except NoOffersApplied as e:
    print("Custom Error:", e)
```

```python
#task4
def calculate_average_rating(total_rating, num_reviews):
    try:
        result = total_rating / num_reviews
        print("Average Rating:", result)
    except ZeroDivisionError:
        print("Error: No reviews found!")
    finally:
        print("Thank you for using the calculator")

calculate_average_rating(500, 0)
```

```python
#task5
def safe_divide_for_zomato(bill_amount, num_people):
    try:
        result = bill_amount / num_people
    except ZeroDivisionError:
        print("Error: Number of people cannot be zero!")
    else:
        print("Each person pays: ₹", result)
    finally:
        print("Split calculation done")

safe_divide_for_zomato(800, 4)
safe_divide_for_zomato(800, 0)
```
