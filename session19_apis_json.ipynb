```python
# Task 1: Fetch posts from a public API and print the first 5 titles
import requests

url = "https://jsonplaceholder.typicode.com/posts"
response = requests.get(url)

# Check if the request was successful
if response.status_code == 200:
    posts = response.json()  # Parse JSON response into a Python list
    
    print("--- First 5 Post Titles ---")
    for index, post in enumerate(posts[:5], start=1):
        print(f"{index}. {post['title']}")
else:
    print(f"Failed to fetch data. Status Code: {response.status_code}")
```

```python
# Task 2: Create a Zomato-style restaurant dictionary and convert to JSON
import json

# Python Dictionary representing a restaurant object
restaurant_data = {
    "name": "The Spice Route",
    "location": "Connaught Place, New Delhi",
    "cuisines": ["North Indian", "Mughlai", "South Indian"],
    "rating": 4.5,
    "is_open": True
}

# Convert the dictionary to a formatted JSON string
# indent=4 makes it beautifully scannable and organized
json_string = json.dumps(restaurant_data, indent=4)

print("--- Serialized JSON Output String ---")
print(json_string)
```

```python
# Task 3: Send a POST request to add data to an endpoint
import requests

url = "https://jsonplaceholder.typicode.com/posts"

# Data payload to send to the server
playlist_payload = {
    "title": "My Chill Vibes Playlist",
    "userId": 42,
    "body": "A curated collection of lo-fi and acoustic tracks for coding."
}

# Using json= automatically stringifies the payload and attaches JSON headers
response = requests.post(url, json=playlist_payload)

print("--- POST Request Verification ---")
print(f"Response Status Code: {response.status_code} (Created)")
print("\n--- JSON Response From Server ---")
print(response.json())
```

```python
# Task 4: Modify GET request using query parameters to filter data
import requests

url = "https://jsonplaceholder.typicode.com/posts"

# Defining query parameters (?userId=2)
query_params = {"userId": 2}

response = requests.get(url, params=query_params)

if response.status_code == 200:
    filtered_posts = response.json()
    
    print("--- Post IDs matching userId = 2 ---")
    for post in filtered_posts:
        print(f"Post ID: {post['id']} (Title: {post['title'][:30]}...)")
else:
    print(f"Request failed with status: {response.status_code}")
```

```python
# Task 5: Sending a GET request containing custom HTTP authorization headers
import requests

# httpbin.org/headers returns the headers sent by the client for confirmation
url = "https://httpbin.org/headers"

# Setting up custom dictionary configuration for HTTP headers
custom_headers = {
    "Authorization": "Bearer YOUR_TOKEN_HERE",
    "User-Agent": "PythonAssignmentBot/1.0"
}

response = requests.get(url, headers=custom_headers)

print("--- Header Request Verification ---")
print(f"Status Code: {response.status_code}")
print("\n--- Server Received Headers ---")
print(response.text)
```
